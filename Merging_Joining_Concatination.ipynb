**How to merege**

```python
# DataFrame 1: Employees information
employees = pd.DataFrame({
    'employee_id': [1,2,3,4,5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1,2,3,6,7],
    'salary': [60000,80000,65000,70000,90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries)  # here Pandas automatically uses how='inner' and identifies the common column employee_id in both DataFrames.
```

```python
pd.merge(employees,salaries, on = 'employee_id')   # by using ---on--- we tell merge according to this 
```

```python
pd.merge(employees,salaries, on = 'employee_id', how = 'inner')   # how='inner' means keep only the rows that have matching values in both DataFrames
```

```python
pd.merge(employees,salaries, on = 'employee_id', how = 'outer')   # how='outer' means keep all rows from both DataFrames, whether their IDs match or not
```

```python
pd.merge(employees,salaries, on = 'employee_id', how = 'left')  # how='left' means keep all rows from the left DataFrame and add matching data from the right DataFrame
# employees is the left DataFrame, and salaries is the right DataFrame
```

```python
pd.merge(employees,salaries, on = 'employee_id', how = 'right')  # how='right' means keep all rows from the right DataFrame and add matching data from the left DataFrame
# employees is the left DataFrame, and salaries is the right DataFrame
```

**Concatination of 2 Data frame**

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2]) # this is concatinate base on the column
```

```python
pd.concat([df2,df1]) # this is concatinate base on the column
```

```python
pd.concat([df1,df2], axis = 1) # this is concatinate base on the row
```

**Joining 2 Data Frames**

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df1.join(df2, how = 'outer')
```
