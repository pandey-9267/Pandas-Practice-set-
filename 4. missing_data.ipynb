```python
import numpy as np
import pandas as pd
```

```python
# Finding Missing Data
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
# Removing missing data
df
```

```python
df.dropna()
```

```python
df.dropna(thresh = 3)  # this give the only those rows which have minimun 3 non-NaN value
```

```python
# Filling the missing data
df
```

```python
df.fillna(0)  # this is replace all the NaN value with any number 
```

```python
val = {'A':100,'B':200,'C':300,'D':400}
df.fillna(value = val)         # by this we can give each columns a different values 
```

```python
df.fillna(df.mean())  # for replacing the NaN with the columns wise mean of the whole 
```
