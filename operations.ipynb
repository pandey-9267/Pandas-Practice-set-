```python
def square (x):
    return x ** 2
```

```python
df1['B'].apply(square)       # this did not change the main df1 
```

```python
df1 # see here that data is not update still give the old data 
```

```python
# fir changing the main df1 we have to store inside the selected column
df1['B'] = df1['B'].apply(square)
```
