```python
# How to create a data frame

data = {
    'Name' : ['Ashraf','Nikhil','Dilip','Abhishek'],
    'Age' : [23,24,25,26],
    'City' : ['Punjab','Bihar','Goa','Delhi'],
    'Salary' : [25000,35222,25001,15666]
}
df1 = pd.DataFrame(data)
```
